# Step 0-B: 벽 설치 가이드

## 1. 액션 번호를 좌표로 변환 (`_action_to_wall`)
벽은 교차점(intersection)에 놓입니다. 9x9 보드에는 8x8 = 64개의 교차점이 있습니다.
가로벽(0) 64개 + 세로벽(1) 64개 = 총 128개의 벽 액션(12~139).

- `wall_index = action - 12` (0~127)
- `orientation = wall_index // 64` (0 또는 1)
- `r = (wall_index % 64) // 8`
- `c = (wall_index % 64) % 8`



## 2. 겹침과 교차 검사 (`_is_valid_wall`)

벽 좌표 `(r, c, d)`가 들어왔을 때:
1. **범위 검사**: `0 <= r < 8` 이고 `0 <= c < 8` 이어야 합니다.
2. **완전 겹침**: `(r, c, d)`가 `walls_placed`에 있으면 안 됨.
3. **절반 겹침**: 
   - `d == 0` (가로)일 때, 좌우 `(r, c-1, 0)` 와 `(r, c+1, 0)` 가 있으면 안 됨.
   - `d == 1` (세로)일 때, 상하 `(r-1, c, 1)` 와 `(r+1, c, 1)` 가 있으면 안 됨.
4. **십자 교차**: `d == 0`일 때 `(r, c, 1)`이 있거나, `d == 1`일 때 `(r, c, 0)`이 있으면 뚫고 지나가는 십자가(┼) 모양이 되므로 금지.


## 3. 길 막기 검사 (BFS, `_has_path`)
너비 우선 탐색(BFS)을 사용해 해당 플레이어의 현재 위치에서 목표 행(`goal_rows[player]`)까지 도달할 수 있는지 검사합니다.

```python
queue = deque([시작_좌표])
visited = set([시작_좌표])

while queue:
    r, c = queue.popleft()
    if r == 목표_행:
        return True
        
    # 상하좌우(MOVE_DELTA)를 순회
    for dr, dc in MOVE_DELTA.values():
        nr, nc = r + dr, c + dc
        if 보드 안이고, (r,c)->(nr,nc) 사이 벽이 없고, (nr,nc) 방문 안했으면:
            visited.add((nr, nc))
            queue.append((nr, nc))

return False # 큐가 빌 때까지 못 찾으면 길 막힌 것!
```
