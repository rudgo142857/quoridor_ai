# Step 0-C: 관점 정규화 가이드

## 1. 관점 정규화가 필요한 이유
우리가 만든 쿼리도 환경에서 Player 0은 맨 아래(8)에서 위(0)로 올라가고, Player 1은 맨 위(0)에서 아래(8)로 내려옵니다.
하지만 인공지능(DQN)에게 이 두 가지 상황을 모두 가르치면 **학습 데이터가 분산되어 학습 속도가 2배로 느려집니다.**

이를 해결하기 위해, 딥마인드의 AlphaZero 등에서는 항상 **'현재 턴인 플레이어'의 관점에서 보드를 180도 뒤집어서** 상태를 입력으로 줍니다. 그러면 AI는 **"나는 항상 맨 아래에서 위로 올라간다"**는 단 한 가지 상황만 학습하면 됩니다!

## 2. 좌표 변환 (180도 회전)
- 9x9 보드에서 `(r, c)`를 180도 뒤집으면 어떻게 될까요?
  `new_r = 8 - r`, `new_c = 8 - c` 가 됩니다.
- 벽 좌표는 어떨까요? 벽은 8x8 교차점에 놓이므로, 교차점 `(r, c)`를 180도 뒤집으면:
  `new_r = 7 - r`, `new_c = 7 - c` 가 됩니다.
