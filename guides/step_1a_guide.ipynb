## 1. 최단 거리 계산 (`get_shortest_path_length`)
이전에 만들었던 `_has_path` 함수는 단순히 "도달 가능한가(True/False)"만 반환했습니다.
이제는 "몇 칸 만에 도달하는가(int)"를 반환하도록 BFS를 살짝 수정해야 합니다.

```python
# 큐에 시작 좌표뿐만 아니라 '현재까지의 거리'도 같이 저장합니다.
queue = deque([(0, r, c)])  # (거리, r, c)
visited = set([(r, c)])

while queue:
    dist, r, c = queue.popleft()
    if r == 목표_행:
        return dist
        
    for dr, dc in MOVE_DELTA.values():
        nr, nc = r + dr, c + dc
        if 이동_가능하고_방문안했으면:
            visited.add((nr, nc))
            queue.append((dist + 1, nr, nc))  # 거리를 1 증가시켜서 넣습니다
```
