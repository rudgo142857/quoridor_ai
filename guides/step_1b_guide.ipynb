# Step 1-B: GreedyWallBot 구현 가이드

## 1. 평가 함수 (Heuristic Function)
`GreedyWallBot`은 체스나 바둑 인공지능이 쓰는 '상태 평가' 기법을 최초로 도입합니다.
현재 상황이 얼마나 유리한지 숫자로 표현해야 에이전트가 어떤 행동을 고를지 결정할 수 있습니다.

쿼리도에서 가장 좋은 휴리스틱은 매우 간단합니다.
**점수 = (상대방이 가야 할 칸 수) - (내가 가야 할 칸 수)**
- 점수가 높을수록 내가 엄청나게 유리한 상황입니다.
- 점수가 낮을수록 내가 불리한 상황입니다.

## 2. 행동 선택 로직 (`act`)
`ShortestPathBot`은 `a < 12`인 이동 액션만 골라서 검사했습니다.
하지만 `GreedyWallBot`은 **벽 설치를 포함한 모든 행동(`legal_moves`)**을 하나씩 가상으로 실행해봅니다.
그리고 방금 만든 `evaluate_state` 함수로 점수를 매긴 뒤, 가장 높은 점수를 주는 행동 하나를 최종 선택합니다.
