## 1. 엡실론-그리디 (Epsilon-Greedy) 탐색
강화학습 에이전트는 무작정 현재 가장 좋아 보이는 행동만 하면 안 됩니다. 더 좋은 숨겨진 길을 찾기 위해 이따금씩 '무작위 탐색(Exploration)'을 해야 합니다.
- `random.random() < self.epsilon`: 이면 무작위로 행동 (`random.choice(legal_moves)`)
- 그렇지 않다면: Q값이 가장 큰 행동을 선택

```python
if random.random() < self.epsilon:
    return random.choice(legal_moves)
else:
    # q_values 딕셔너리에서 가장 값이 큰 행동(Key) 찾기
    return max(q_values, key=q_values.get)
```

## 2. 제로섬 게임의 벨만 업데이트 (Zero-Sum Bellman Update)
쿼리도는 한 명이 이기면 다른 한 명은 지는 완벽한 제로섬(Zero-Sum) 게임입니다.
즉, 내가 한 턴을 진행하고 난 뒤의 다음 상태(`next_state`)는 상대방의 턴입니다.
상대방이 그 상태에서 누릴 수 있는 최대 이익(`max_next_q`)은 곧 나에게는 손해(-)가 됩니다.

따라서 업데이트 공식의 타겟(Target) 값은 다음과 같이 부호가 뒤집힙니다:
- 일반 게임: `Target = Reward + (gamma * max_next_q)`
- 턴제 제로섬 게임: `Target = Reward - (gamma * max_next_q)`

```python
target = reward - (self.gamma * max_next_q)
self.q_table[state_key][action] = old_q + self.alpha * (target - old_q)
```
